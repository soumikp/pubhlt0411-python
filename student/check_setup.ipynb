```python
# Read a real file and compute a real number.
overdose = pd.read_csv("data/pa_overdose_county_year.csv")

y2023 = overdose[overdose["year"] == 2023].dropna(subset=["overdose_deaths"])
rate = y2023["overdose_deaths"].sum() / y2023["population"].sum() * 100_000

print(f"{len(overdose)} rows, {overdose['county'].nunique()} counties")
print(f"2023 overdose death rate: {rate:.1f} per 100,000")
```

```python
# Draw a figure. If a plot appears below, graphics work.
fig, ax = plt.subplots(figsize=(7, 2.4))

ax.scatter(y2023["rate_per_100k"], [0] * len(y2023),
           s=45, color="#003594", alpha=0.55, edgecolor="white")
ax.set_yticks([])
ax.set_xlabel("Overdose deaths per 100,000 residents, 2023")
ax.set_title("Each dot is one Pennsylvania county")
for side in ("left", "right", "top"):
    ax.spines[side].set_visible(False)
plt.tight_layout()
plt.show()
```
